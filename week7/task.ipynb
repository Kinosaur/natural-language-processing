```python
from gensim.models import Word2Vec

sentences = [
    ['I', "love", "this", "movie"],
    ['This', "movie", "is", "great"],
    ['the', 'food', 'was', 'delicious'],
    ["bad", "service", "ruined", "the", "experience"]
]

model = Word2Vec(sentences, vector_size=50, window=3, min_count=1)

# Show the words most similar to "movie" in the tiny toy corpus
similar = model.wv.most_similar('movie')
print(similar)
```

```python
import os
import gensim.downloader as api

base = "/Users/kaungkhantlin/Developer/2_2025/NLP/week7/models"
os.makedirs(base, exist_ok=True)
os.environ["GENSIM_DATA_DIR"] = base  # tell gensim where to cache

model = api.load("word2vec-google-news-300")  # ~1.6GB download, ~3.4GB expanded
model.save(os.path.join(base, "GoogleNews-vectors-negative300.gensim"))
```
